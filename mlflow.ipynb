```python
# --- 1. Imports Necessários ---
import mlflow
import mlflow.sklearn
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
import warnings

warnings.filterwarnings('ignore')

print("Bibliotecas importadas com sucesso.")
```

```python
# --- 2. Preparação dos Dados ---
dados = load_diabetes()
X = dados.data
y = dados.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
print("Dados carregados e separados (treino/teste).")

# --- 3. Configuração do Experimento ---
experiment_name = "Seminario_Demo_Diabetes_v2"
mlflow.set_experiment(experiment_name)
print(f"MLflow configurado para o experimento: '{experiment_name}'")
print(f"Abrir a interface so usar o mlflow UI")
```

REGRESSÃO LINEAR

```python
with mlflow.start_run(run_name="Regressao_Linear"):
    print("\n Iniciando Run: Regressão Linear")

    model = LinearRegression()

    # Não tem hiperparâmetros relevantes para logar
    mlflow.log_param("model_type", "LinearRegression")

    # Treinamento
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    # Métrica
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mlflow.log_metric("rmse", rmse)

    # Artefato: o modelo salvo
    mlflow.sklearn.log_model(model, name = "modelo_linear")

    print(f"Run concluída: RMSE = {rmse:.4f}")
```

ÁRVORE DE DECISÃO

```python
with mlflow.start_run(run_name="Arvore_Decisao"):
    print("\nIniciando Run: Árvore de Decisão")

    model = DecisionTreeRegressor(max_depth=5, random_state=42)

    # Parâmetros
    mlflow.log_param("model_type", "DecisionTreeRegressor")
    mlflow.log_param("max_depth", 5)

    # Treinamento
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    # Métrica
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mlflow.log_metric("rmse", rmse)

    # Artefato
    mlflow.sklearn.log_model(model, name = "modelo_arvore")

    print(f"Run concluída: RMSE = {rmse:.4f}")
```

RANDOM FOREST

```python
with mlflow.start_run(run_name="Random_Forest"):
    print("\n Iniciando Run: Random Forest")

    model = RandomForestRegressor(n_estimators=50, random_state=42)

    # Parâmetros
    mlflow.log_param("model_type", "RandomForestRegressor")
    mlflow.log_param("n_estimators", 50)

    # Treinamento
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    # Métrica
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    mlflow.log_metric("rmse", rmse)

    # Artefato
    mlflow.sklearn.log_model(model, "modelo_random_forest")

    print(f"Run concluída: RMSE = {rmse:.4f}")
```
